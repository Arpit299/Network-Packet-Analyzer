# Network Packet Analyzer
This notebook demonstrates a basic network packet analyzer using the `scapy` library to capture and inspect network traffic.

```python
!pip install scapy
```

```python
from scapy.all import sniff, IP, TCP, UDP

def packet_callback(packet):
    if IP in packet:
        ip_src = packet[IP].src
        ip_dst = packet[IP].dst
        proto = "TCP" if TCP in packet else "UDP" if UDP in packet else "Other"

        print(f"[+] {proto} Packet: {ip_src} -> {ip_dst}")

print("Starting packet sniffer (capturing 10 packets)...")
sniff(prn=packet_callback, count=10)
```
